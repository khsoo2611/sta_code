```python
from pathlib import Path

import pandas as pd

DATA_DIR = Path("data")
```

Physicochemical Properties of Protein Tertiary Structure (UCI). The processed file
keeps the features F1-F9 and the response RMSD.

```python
raw = pd.read_csv(DATA_DIR / "CASP.csv")
print(raw.shape, "missing:", raw.isna().sum().sum())
raw.head(3).T
```

```python
FEATURES = [f"F{i}" for i in range(1, 10)]

data = raw[FEATURES + ["RMSD"]].dropna()
assert len(data) == 45730
data.to_csv(DATA_DIR / "CASP_processed.csv", index=False)
data.shape
```
