## Protein

```python
labeled_sizes = range(100, 1001, 100)
N = 5000
reps = 100
seed = 202607196
jobs = 12
estimators = [
    "ata_sbf",
    "ata_csbf",
    "lp_nw_theory",
    "lp_ll_theory",
    "krr_m",
    "krr_r",
    "ease_nw",
    "ease_ll",
    "ease_m",
    "ease_r",
    "pi",
]

K = 5
r = 2
cv_folds = 10

summary_protein = {}
records_protein = {}
for n in labeled_sizes:
    H = n // 5
    summary_protein[n], records_protein[n] = run_realdata(
        "protein",
        estimators=estimators,
        n=n,
        N=N,
        reps=reps,
        seed=seed,
        K=K,
        r=r,
        H=H,
        cv_folds=cv_folds,
        jobs=jobs,
        run_name=f"protein_n{n}_N{N}",
    )
```

```python
import pandas as pd

table_protein = pd.concat(summary_protein.values(), ignore_index=True).pivot(
    index="estimator_id", columns="n", values="ratio"
).round(4)
table_protein
```
